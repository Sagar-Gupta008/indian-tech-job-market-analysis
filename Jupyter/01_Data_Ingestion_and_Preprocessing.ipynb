## In this section, we will add new columns that are necessary for us in future 

## Feature 1: Salary Range

```python
import numpy as np

conditions = [
    (df_clean['salary_midpoint_lpa'] == 0),
    (df_clean['salary_midpoint_lpa'] < 5),
    (df_clean['salary_midpoint_lpa'] >= 5) & (df_clean['salary_midpoint_lpa'] < 10),
    (df_clean['salary_midpoint_lpa'] >= 10) & (df_clean['salary_midpoint_lpa'] < 20),
    (df_clean['salary_midpoint_lpa'] >= 20)
]

choices = [
    "Not Disclosed",
    "Low (Below 5 LPA)",
    "Medium (5–10 LPA)",
    "High (10–20 LPA)",
    "Premium (20+ LPA)"
]

df_clean['salary_range'] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
df_clean['salary_range'].value_counts()
```

## Business Question
## Which salary band has the highest number of job openings?
## Ans High (10-20 LPA)

## Feature 2: Experience Span

```python
df_clean['experience_span'] = (
    df_clean['experience_max_yrs'] -
    df_clean['experience_min_yrs']
)
```

```python
df_clean['experience_span'].describe()
```

## Data Quality Investigation - Experience Span

## objective : Identify why some records have a negative experience span.

## Step 1: Find the Problematic Records

```python
negative_exp = df_clean[df_clean['experience_span'] < 0]

negative_exp
```

## Step 2: Check How Many Records Are Affected

```python
negative_exp.shape
```

```python
negative_exp[
    [
        'job_id',
        'job_title',
        'experience_min_yrs',
        'experience_max_yrs',
        'experience_raw'
    ]
]
```

## Root Cause Analysis - Negative Experience Span

## Determine whether these 90 records are:

## - Scraping errors
## - Data entry errors
## - Legitimate values that need different handling

## Step 3: Display the Problematic Records

```python
negative_exp[
    [
        'job_id',
        'job_title',
        'experience_raw',
        'experience_min_yrs',
        'experience_max_yrs'
    ]
].head(20)
```

## Step 4: Check Unique Experience Formats

```python
negative_exp['experience_raw'].value_counts()
```

## Instead of removing thos 90 records, Flag these records as invalid and exclude them from analyses that depend on experience.

```python
walkin_mask = df_clean['experience_span'] < 0
df_clean['experience_data_valid'] = ~walkin_mask
```

## preserve the original values instead of overwriting them.

```python
df_clean['experience_issue'] = np.where(
    df_clean['experience_span'] < 0,
    'Walk-in Date Parsed as Experience',
    'Valid'
)
```

### Experience Data Validation

During data validation, 90 records (0.39% of the dataset) were identified with negative experience spans. Investigation showed these were primarily walk-in job postings where the web scraper incorrectly interpreted interview dates (e.g., "13-Jun") as experience values. Rather than deleting these records or fabricating corrected values, they were flagged using an `experience_data_valid` indicator and excluded from experience-based analyses.

## Feature 4: Job Age Category

```python
conditions = [
    df_clean['days_since_posted'] <= 7,
    (df_clean['days_since_posted'] > 7) & (df_clean['days_since_posted'] <= 30),
    df_clean['days_since_posted'] > 30
]

choices = [
    "Recent (0–7 Days)",
    "Active (8–30 Days)",
    "Older (30+ Days)"
]

df_clean['job_age_category'] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
df_clean['job_age_category'].value_counts()
```

## Business Question
## Are most job postings recent or old?
## Ans Most of the jobs are active(8-30 days) followed by Recent jobs(0-7 days)

## Feature 5: Multi-Location Jobs

```python
df_clean['is_multi_location'] = (
    df_clean['location']
    .str.contains(',', na=False)
)
```

```python
df_clean['is_multi_location'].value_counts()
```

## Business Question
## How many jobs are open across multiple locations?
## Ans 6455

## Feature 6: Company Rating Category

```python
conditions = [
    df_clean['company_rating'] < 3,
    (df_clean['company_rating'] >= 3) & (df_clean['company_rating'] < 4),
    (df_clean['company_rating'] >= 4) & (df_clean['company_rating'] < 4.5),
    df_clean['company_rating'] >= 4.5
]

choices = [
    "Below Average",
    "Good",
    "Very Good",
    "Excellent"
]

df_clean['company_rating_category'] = np.select(
    conditions,
    choices,
    default="Not Rated"
)
```

```python
df_clean['company_rating_category'].value_counts()
```

## Business Question
